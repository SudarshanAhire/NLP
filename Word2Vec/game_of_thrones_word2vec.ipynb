```python
import kagglehub

# Download latest version
data = kagglehub.dataset_download("khulasasndh/game-of-thrones-books")

print("Path to dataset files:", data)
```

```python
import nltk
from nltk import sent_tokenize
from gensim.utils import simple_preprocess
from nltk.corpus import stopwords

nltk.download('punkt_tab')
nltk.download('stopwords')

stop_words = set(stopwords.words('english'))

story = []
dataset_path = '/root/.cache/kagglehub/datasets/khulasasndh/game-of-thrones-books/versions/1'

for filename in os.listdir(dataset_path):

    f = open(os.path.join(dataset_path, filename), 'r', errors='ignore')
    corpus = f.read()
    raw_sent = sent_tokenize(corpus)

    for sent in raw_sent:
        # Preprocess the sentence into a list of words
        processed_words = simple_preprocess(sent)

        # Filter out the stopwords
        filtered_words = [word for word in processed_words if word not in stop_words]

        # Append the cleaned sentence words to the story
        story.append(filtered_words)

    f.close()
```

```python
story
```

```python
model = gensim.models.Word2Vec(
    window=10,
    min_count=2
)
```

```python
model.build_vocab(story)
```

```python
model.train(story, total_examples=model.corpus_count, epochs=model.epochs)
```
